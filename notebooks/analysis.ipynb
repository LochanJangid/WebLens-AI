## HTML parsing

```python
from bs4 import BeautifulSoup

html_doc = """
<html>
  <body>
    <h1>this is the website title brooo</h1>
    <p class="description">This is a paragraph of text.</p>
    <a href="https://lochan.vercel.com" id="link1">go for portfolio visit</a>
  </body>
</html>
"""

soup = BeautifulSoup(html_doc, 'html.parser')
soup.find('h1').text, soup.find('a')['href']
```
